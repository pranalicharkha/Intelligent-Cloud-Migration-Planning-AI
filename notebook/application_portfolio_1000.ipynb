```python
from faker import Faker
import pandas as pd
import random

fake = Faker()
```

```python
num_apps = 1000

applications = []

for i in range(num_apps):

    app = {
        "application_id": f"APP{i+1:03d}",
        "application_name": fake.bs(),
        "cpu_usage": round(random.uniform(0.1, 1.0), 2),
        "memory_usage": round(random.uniform(0.1, 1.0), 2),
        "age_years": random.randint(1, 15),
        "criticality": random.choice(["Low", "Medium", "High"]),
        "compliance_flag": random.choice([0, 1])
    }

    applications.append(app)

df = pd.DataFrame(applications)

df.head()
```

```python
app_ids = df["application_id"].tolist()

dependency_ids = []

for app_id in app_ids:

    possible_dependencies = [
        x for x in app_ids
        if x != app_id
    ]

    num_dependencies = random.randint(0, 5)

    dependencies = random.sample(
        possible_dependencies,
        min(num_dependencies, len(possible_dependencies))
    )

    dependency_ids.append(",".join(dependencies))

df["dependency_ids"] = dependency_ids

# Replace empty dependency values with 0
df["dependency_ids"] = df["dependency_ids"].replace("", 0)

df["dependency_count"] = df["dependency_ids"].apply(
    lambda x: 0 if x == 0 else len(x.split(","))
)

df.head()
```

```python
df.to_csv(
    "application_portfolio.csv",
    index=False
)
```
