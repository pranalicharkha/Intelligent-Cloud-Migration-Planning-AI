```python
numeric_cols = [
    "cpu_rate",
    "canonical_memory_usage",
    "assigned_memory_usage",
    "unmapped_page_cache",
    "total_page_cache",
    "max_memory_usage",
    "mean_disk_io_time",
    "mean_local_disk_space",
    "max_cpu_rate",
    "sampled_cpu_usage"
]

negative_values = (migration_data[numeric_cols] < 0).sum()

print(negative_values)
```

```python
task_summary = migration_data.groupby("task_id").agg(
    avg_cpu=("cpu_rate", "mean"),
    max_cpu=("max_cpu_rate", "max"),
    avg_memory=("canonical_memory_usage", "mean"),
    max_memory=("max_memory_usage", "max"),
    avg_disk_io=("mean_disk_io_time", "mean"),
    duration_seconds=("duration_seconds", "sum")
).reset_index()

print(task_summary.head())
print("\nShape:", task_summary.shape)
```

```python
task_lifetime = migration_data.groupby("task_id").agg(
    first_start=("start_time", "min"),
    last_end=("end_time", "max")
).reset_index()

task_lifetime["lifetime_seconds"] = (
    task_lifetime["last_end"] - task_lifetime["first_start"]
) / 1_000_000

print(task_lifetime.head())
print("\nLifetime statistics:")
print(task_lifetime["lifetime_seconds"].describe())
```

```python
task_summary = task_summary.drop(columns=["duration_seconds"])

task_summary = task_summary.merge(
    task_lifetime[["task_id", "lifetime_seconds"]],
    on="task_id",
    how="left"
)

print(task_summary.head())
print("\nShape:", task_summary.shape)
```

```python
cpu_variability = migration_data.groupby("task_id").agg(
    cpu_std=("cpu_rate", "std")
).reset_index()

task_summary = task_summary.merge(
    cpu_variability,
    on="task_id",
    how="left"
)

print(task_summary.head())
print("\nCPU variability statistics:")
print(task_summary["cpu_std"].describe())
```

```python
task_summary["cpu_std"] = task_summary["cpu_std"].fillna(0)

print("Missing cpu_std:", task_summary["cpu_std"].isna().sum())
print(task_summary["cpu_std"].describe())
```

```python
memory_variability = migration_data.groupby("task_id").agg(
    memory_std=("canonical_memory_usage", "std")
).reset_index()

task_summary = task_summary.merge(
    memory_variability,
    on="task_id",
    how="left"
)

task_summary["memory_std"] = task_summary["memory_std"].fillna(0)

print(task_summary.head())
print("\nMemory variability statistics:")
print(task_summary["memory_std"].describe())
```

```python
observation_counts = migration_data.groupby("task_id").size().reset_index(
    name="num_observations"
)

task_summary = task_summary.merge(
    observation_counts,
    on="task_id",
    how="left"
)

print(task_summary.head())
print("\nObservation count statistics:")
print(task_summary["num_observations"].describe())
```

```python
ml_features = [
    "avg_cpu",
    "max_cpu",
    "avg_memory",
    "max_memory",
    "avg_disk_io",
    "lifetime_seconds",
    "cpu_std",
    "memory_std",
    "num_observations",
    "num_events",
    "cpu_capacity",
    "memory_capacity",
    "priority",
    "scheduling_class",
    "cpu_utilization_ratio",
    "memory_utilization_ratio",
    "cpu_headroom",
    "memory_headroom",
    "overall_resource_utilization"
]

X = task_summary[ml_features].copy()

print("ML feature matrix shape:", X.shape)
print("\nNumber of features:", len(ml_features))
print("\nFeatures:")
print(ml_features)
```

```python
correlation_matrix = X.corr()

print(
    correlation_matrix[
        ["cpu_utilization_ratio",
         "memory_utilization_ratio",
         "overall_resource_utilization"]
    ].round(3)
)
```

```python
corr_pairs = (
    correlation_matrix
    .where(
        ~__import__("numpy").tril(
            __import__("numpy").ones(correlation_matrix.shape),
            k=0
        ).astype(bool)
    )
    .stack()
    .sort_values(key=abs, ascending=False)
)

print("Feature pairs with |correlation| >= 0.90:\n")

print(
    corr_pairs[corr_pairs.abs() >= 0.90]
)
```

```python
high_corr = corr_pairs[corr_pairs.abs() >= 0.90].reset_index()

high_corr.columns = [
    "feature_1",
    "feature_2",
    "correlation"
]

print(high_corr)
```
